```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ai_student_impact_dataset.csv')
```

```python
subscription_mapping = {True: 1, False: 0}
df['Paid_Subscription'] = df['Paid_Subscription'].map(subscription_mapping)
```

```python
## Which Major Category has the highest number of students?

category_with_highest_students = df['Major_Category'].value_counts().idxmax()
category_with_highest_students
```

```python
df['Major_Category'].value_counts()
```

```python
## Which Year of Study uses GenAI the most?

year_of_study_with_highest_genai_usage = df.groupby('Year_of_Study')['Weekly_GenAI_Hours'].sum().idxmax()
print(year_of_study_with_highest_genai_usage)

year_of_study_with_lowest_genai_usage = df.groupby('Year_of_Study')['Weekly_GenAI_Hours'].sum().idxmin()
print(year_of_study_with_lowest_genai_usage)
```

```python
df.head()
```

```python
## Which year (Freshman, Sophomore, Junior, Senior) has the highest GPA improvement? GPA improvement = Final GPA - Initial GPA

df['GPA_Improvement'] = df['Post_Semester_GPA'] - df['Pre_Semester_GPA']

year_with_highest_gpa_improvement = df.groupby('Year_of_Study')['GPA_Improvement'].mean().idxmax()
print(year_with_highest_gpa_improvement)
```

```python
## How many hours per week does the average student spend using GenAI?

average_genai_hours = df.groupby('Year_of_Study')['Weekly_GenAI_Hours'].mean().groupby('Year_of_Study').mean()
print(average_genai_hours)
```

```python
## Which Primary Use Case is the most common?

most_common_use_case = df['Primary_Use_Case'].value_counts().idxmax()
print(most_common_use_case)
```

```python
## Which use case is associated with the largest GPA improvement?

use_case_with_largest_gpa_improvement = df.groupby('Primary_Use_Case')['GPA_Improvement'].mean().idxmax()
print(use_case_with_largest_gpa_improvement)
```

```python
## Does spending more hours with GenAI generally improve GPA?

correlation = df['Weekly_GenAI_Hours'].corr(df['GPA_Improvement'])
print(f"Correlation between Weekly GenAI Hours and GPA Improvement: {correlation}")
```

```python
## Which students improved the most this semester?

top_improvers = df.nlargest(5, 'GPA_Improvement')
top_improvers
```

```python
## Are students paying for AI actually getting better results?

paid_students = df[df['Paid_Subscription'] == 1]
unpaid_students = df[df['Paid_Subscription'] == 0]

print(f"Average GPA Improvement for Paid Students: {paid_students['GPA_Improvement'].mean()}")
print(f"Average GPA Improvement for Unpaid Students: {unpaid_students['GPA_Improvement'].mean()}")
```

```python
## Who is becoming most dependent on AI? Analyze student by: Major, Year, Prompt Skill, Use Case

more_dependent_major_category = df.groupby('Major_Category')['Weekly_GenAI_Hours'].mean().idxmax()
print(f"Major Category with highest average GenAI usage: {more_dependent_major_category}")
```

```python
## What factors seem to increase burnout? Investigate relationships with: Weekly_GenAI_Hours, Traditional_Study_Hours, Anxiety_Level_During_Exams

burnout_correlation_genai = df['Weekly_GenAI_Hours'].corr(df['Anxiety_Level_During_Exams'])
burnout_correlation_study = df['Traditional_Study_Hours'].corr(df['Anxiety_Level_During_Exams'])
print(f"Correlation between GenAI usage and burnout: {burnout_correlation_genai}")
print(f"Correlation between traditional study hours and burnout: {burnout_correlation_study}")
```
